# 📡 Notebook 01: Data Collection

**Objective:** Fetch historical air pollution and meteorological data from the Open-Meteo API.

- **Air Quality API:** PM2.5, PM10, NO₂, SO₂, CO, O₃
- **Weather API:** Temperature, Humidity, Wind Speed, Pressure, Dew Point
- **Target City:** Delhi, India (28.6139°N, 77.2090°E)
- **Period:** 2022-01-01 to 2025-12-31 (4 years, hourly)

---

```python
import sys
import os
sys.path.insert(0, os.path.abspath('..'))

import pandas as pd
import warnings
warnings.filterwarnings('ignore')

from src.data.data_loader import load_config
from src.data.pollution_fetch import fetch_pollution_data
from src.data.weather_fetch import fetch_weather_data
from src.data.data_merger import merge_pollution_weather

# Load configuration
config = load_config('../config/config.yaml')
print(f"City: {config['data']['city']}")
print(f"Coordinates: ({config['data']['latitude']}, {config['data']['longitude']})")
print(f"Period: {config['data']['start_date']} to {config['data']['end_date']}")
```

## 1.1 Fetch Air Pollution Data

```python
# Fetch pollution data from Open-Meteo Air Quality API
pollution_df = fetch_pollution_data(
    latitude=config['data']['latitude'],
    longitude=config['data']['longitude'],
    start_date=config['data']['start_date'],
    end_date=config['data']['end_date'],
    pollutants=config['data']['pollutants'],
    save_path='../' + config['data']['raw_pollution_path']
)

print(f"\nShape: {pollution_df.shape}")
pollution_df.head()
```

## 1.2 Fetch Meteorological Data

```python
# Fetch weather data from Open-Meteo Historical Weather API
weather_df = fetch_weather_data(
    latitude=config['data']['latitude'],
    longitude=config['data']['longitude'],
    start_date=config['data']['start_date'],
    end_date=config['data']['end_date'],
    variables=config['data']['weather_variables'],
    save_path='../' + config['data']['raw_weather_path']
)

print(f"\nShape: {weather_df.shape}")
weather_df.head()
```

## 1.3 Merge Pollution + Weather Data

```python
# Merge both datasets on timestamp
merged_df = merge_pollution_weather(
    pollution_df, weather_df,
    save_path='../' + config['data']['merged_path']
)

print(f"\nMerged Shape: {merged_df.shape}")
print(f"Columns: {list(merged_df.columns)}")
merged_df.head()
```
