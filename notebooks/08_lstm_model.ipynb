# 🔄 Notebook 08: LSTM Model Training

**Objective:** Train the Long Short-Term Memory network for AQI category classification.

Architecture: LSTM(128) → Dropout → LSTM(64) → Dropout → Dense → Softmax

---

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))

import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

from src.data.data_loader import load_config
from src.preprocessing.sequence_generator import load_sequences
from src.training.train_lstm import train_lstm

config = load_config('../config/config.yaml')
X_train, X_test, y_train, y_test = load_sequences(
    '../' + config['data']['sequences_path']
)
print(f"X_train: {X_train.shape}, X_test: {X_test.shape}")
```

## 8.1 Build and Train LSTM

```python
lstm_model, lstm_history = train_lstm(
    X_train, y_train, X_test, y_test,
    config=config['training'],
    model_config=config['models']['lstm'],
    checkpoint_dir='../models'
)
```

## 8.2 Training History

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(lstm_history.history['loss'], label='Train Loss')
axes[0].plot(lstm_history.history['val_loss'], label='Val Loss')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Loss')
axes[0].set_title('LSTM — Loss')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

axes[1].plot(lstm_history.history['accuracy'], label='Train Acc')
axes[1].plot(lstm_history.history['val_accuracy'], label='Val Acc')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Accuracy')
axes[1].set_title('LSTM — Accuracy')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.suptitle('LSTM Model Training History', fontsize=14)
plt.tight_layout()
plt.savefig('../results/lstm_training_history.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 8.3 Quick Evaluation

```python
from src.evaluation.metrics import print_model_metrics
from src.evaluation.confusion_matrix import plot_confusion_matrix

y_pred_lstm = np.argmax(lstm_model.predict(X_test, verbose=0), axis=1)

lstm_metrics = print_model_metrics('LSTM', y_test, y_pred_lstm)

plot_confusion_matrix(y_test, y_pred_lstm, 'LSTM',
                      save_path='../results/lstm_confusion_matrix.png')

np.save('../artifacts/y_pred_lstm.npy', y_pred_lstm)
print('LSTM predictions saved.')
```
