# ⚡ Notebook 09: Hybrid CNN-LSTM Model Training

**Objective:** Train the core Hybrid CNN-LSTM model combining convolutional feature extraction with LSTM temporal modelling.

Architecture: Conv1D → BN → ReLU → Conv1D → BN → ReLU → MaxPool → LSTM(128) → LSTM(64) → Dense → Softmax

---

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))

import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

from src.data.data_loader import load_config
from src.preprocessing.sequence_generator import load_sequences
from src.training.train_cnn_lstm import train_cnn_lstm

config = load_config('../config/config.yaml')
X_train, X_test, y_train, y_test = load_sequences(
    '../' + config['data']['sequences_path']
)
print(f"X_train: {X_train.shape}, X_test: {X_test.shape}")
```

## 9.1 Build and Train Hybrid CNN-LSTM

```python
hybrid_model, hybrid_history = train_cnn_lstm(
    X_train, y_train, X_test, y_test,
    config=config['training'],
    model_config=config['models']['cnn_lstm'],
    checkpoint_dir='../models'
)
```

## 9.2 Training History

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(hybrid_history.history['loss'], label='Train Loss')
axes[0].plot(hybrid_history.history['val_loss'], label='Val Loss')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Loss')
axes[0].set_title('Hybrid CNN-LSTM — Loss')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

axes[1].plot(hybrid_history.history['accuracy'], label='Train Acc')
axes[1].plot(hybrid_history.history['val_accuracy'], label='Val Acc')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Accuracy')
axes[1].set_title('Hybrid CNN-LSTM — Accuracy')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.suptitle('Hybrid CNN-LSTM Model Training History', fontsize=14)
plt.tight_layout()
plt.savefig('../results/cnn_lstm_training_history.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 9.3 Quick Evaluation

```python
from src.evaluation.metrics import print_model_metrics
from src.evaluation.confusion_matrix import plot_side_by_side_cm

y_pred_hybrid = np.argmax(hybrid_model.predict(X_test, verbose=0), axis=1)

hybrid_metrics = print_model_metrics('Hybrid CNN-LSTM', y_test, y_pred_hybrid)

plot_side_by_side_cm(y_test, y_pred_hybrid, 'Hybrid CNN-LSTM',
                     save_path='../results/cnn_lstm_confusion_matrix.png')

np.save('../artifacts/y_pred_hybrid.npy', y_pred_hybrid)
print('Hybrid CNN-LSTM predictions saved.')
```
