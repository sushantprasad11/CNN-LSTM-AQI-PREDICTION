# 📊 Notebook 03: AQI Calculation

**Objective:** Compute the Air Quality Index using the Indian CPCB standard.

---

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

from src.data.data_loader import load_config, load_csv
from src.preprocessing.aqi_calculator import calculate_aqi_dataframe, AQI_CATEGORIES

config = load_config('../config/config.yaml')
df = load_csv('../' + config['data']['merged_path'])
print(f"Loaded: {df.shape}")
```

## 3.1 Compute AQI and Categories

```python
# Calculate AQI for every timestamp
aqi_df = calculate_aqi_dataframe(df, save_path='../' + config['data']['aqi_path'])
aqi_df[['pm2_5', 'pm10', 'no2', 'so2', 'co', 'o3', 'aqi', 'aqi_category', 'aqi_code']].head(10)
```

## 3.2 AQI Distribution

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# AQI histogram
axes[0].hist(aqi_df['aqi'], bins=50, color='steelblue', edgecolor='white', alpha=0.8)
axes[0].set_xlabel('AQI Value')
axes[0].set_ylabel('Frequency')
axes[0].set_title('AQI Value Distribution')
axes[0].axvline(x=50, color='green', linestyle='--', alpha=0.5, label='Good/Satisfactory')
axes[0].axvline(x=200, color='orange', linestyle='--', alpha=0.5, label='Moderate/Poor')
axes[0].axvline(x=300, color='red', linestyle='--', alpha=0.5, label='Poor/Very Poor')
axes[0].legend()

# Category bar chart
cat_order = ['Good', 'Satisfactory', 'Moderate', 'Poor', 'Very Poor', 'Severe']
colors = ['#009966', '#FFDE33', '#FF9933', '#CC0033', '#660099', '#7E0023']
cat_counts = aqi_df['aqi_category'].value_counts().reindex(cat_order).fillna(0)
axes[1].bar(cat_counts.index, cat_counts.values, color=colors, edgecolor='white')
axes[1].set_xlabel('AQI Category')
axes[1].set_ylabel('Count')
axes[1].set_title('AQI Category Distribution')
plt.xticks(rotation=30)

plt.suptitle('Air Quality Index Analysis — Delhi', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
# AQI over time
fig, ax = plt.subplots(figsize=(16, 5))
ax.plot(aqi_df.index, aqi_df['aqi'], linewidth=0.3, alpha=0.7, color='steelblue')
ax.axhline(y=200, color='orange', linestyle='--', alpha=0.5, label='Moderate/Poor threshold')
ax.axhline(y=300, color='red', linestyle='--', alpha=0.5, label='Very Poor threshold')
ax.set_xlabel('Date')
ax.set_ylabel('AQI')
ax.set_title('AQI Time Series — Delhi (2022-2025)')
ax.legend()
plt.tight_layout()
plt.show()
```
